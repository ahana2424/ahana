```python
import numpy as np
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

print(arr * 4)
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
min_idx = np.argmin(arr)
max_idx = np.argmax(arr)
print(min_idx, max_idx)

print(f"Value at min_idx ({min_idx}): {arr[min_idx]}")
print(f"Value at max_idx ({max_idx}): {arr[max_idx]}")

print(np.min(arr))
print(np.max(arr))

newarray = arr[arr > 20]
print("Values > 20:", newarray)
```

```python
matrix = np.array([[5, 12, 17, 9],
[8, 21, 14, 3],
[6, 13, 19, 11]])

print("Shape:", matrix.shape)
print(f"Rows: {matrix.shape[0]}, Columns: {matrix.shape[1]}")

print("Third column:", matrix[:, 2])

print("Second row:", matrix[1, :])

print("Multiplied by 2:\n", matrix * 2)

print("Mean of each column:", np.mean(matrix, axis=0))

print("Overall mean:", np.mean(matrix))

print("Max value in each row:", np.max(matrix, axis=1))
print("Min value in each column:", np.min(matrix, axis=0))

flat_matrix = matrix.flatten()
print("Flattened shape:", flat_matrix.shape)
print("Flattened array:", flat_matrix)

flat_min_idx = np.argmin(flat_matrix)
flat_max_idx = np.argmax(flat_matrix)
print(f"Flat min index: {flat_min_idx}, Flat max index: {flat_max_idx}")

print("Coordinates of min value:", np.unravel_index(flat_min_idx, matrix.shape))
print("Coordinates of max value:", np.unravel_index(flat_max_idx, matrix.shape))

print("Values > 15:", matrix[matrix > 15])

sorted_matrix = np.sort(matrix, axis=1)
print("Sorted matrix:\n", sorted_matrix)
```

```python
def filter_prime_rows(arr):
    def is_prime(n):
        if n < 2:
            return False
        for i in range(2, int(np.sqrt(n)) + 1):
            if n % i == 0:
                return False
        return True
    
    v_is_prime = np.vectorize(is_prime)
    prime_mask = v_is_prime(arr)
    
    return arr[np.any(prime_mask, axis=1)]

arr_stars = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
print(filter_prime_rows(arr_stars))
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 0::2] = 1
    board[1::2, 1::2] = 1
    return board

def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 1::2] = 1
    board[1::2, 0::2] = 1
    return board

print("checkerboard():\n", checkerboard())
print("\nreverse_checkerboard():\n", reverse_checkerboard())
```

```python
def secondDimmest(stars_array):
    sorted_rows = np.sort(stars_array, axis=1)
    return sorted_rows[:, -2]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print("Stars Matrix:\n", stars)
print("\nSecond Brightest from each cluster:\n", secondDimmest(stars))
```
