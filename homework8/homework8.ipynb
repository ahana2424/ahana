```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

linestyles = ['-', '--', '-.', ':', (0, (3, 1, 1, 1))]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w, ls in zip(amplitude, frequency, linestyles):
    y1 = make_sine_wave(x, A, 1)
    axes[0].plot(x, y1, linestyle=ls, label=f"A = {A}, ω = 1")
    
    y2 = make_sine_wave(x, 1, w)
    axes[1].plot(x, y2, linestyle=ls, label=f"A = 1, ω = {w}")
    
    y3 = make_sine_wave(x, A, w)
    axes[2].plot(x, y3, linestyle=ls, label=f"A = {A}, ω = {w}")

titles = ["Varying Amplitude (ω = 1)", "Varying Frequency (A = 1)", "Varying Both Amplitude & Frequency"]

for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=16)
    ax.set_xlabel("x (radians)", fontsize=14)
    ax.set_ylabel("y = A · sin(ωx)", fontsize=14)
    ax.legend(fontsize=11, loc="upper right")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
