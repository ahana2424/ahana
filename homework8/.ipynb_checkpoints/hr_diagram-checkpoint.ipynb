```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)

print("Data shape:", data.shape)
print(f"Number of stars (rows): {data.shape[0]}")

magnitude = data[:, 0]
temperature = data[:, 1]

min_mag = np.min(magnitude)
max_mag = np.max(magnitude)
brightest_idx = np.argmin(magnitude)
dimmest_idx = np.argmax(magnitude)

print(f"Brightest star (min magnitude): {min_mag} at row index {brightest_idx} (Temp: {temperature[brightest_idx]} K)")
print(f"Dimmest star (max magnitude): {max_mag} at row index {dimmest_idx} (Temp: {temperature[dimmest_idx]} K)")

min_temp = np.min(temperature)
max_temp = np.max(temperature)
coldest_idx = np.argmin(temperature)
hottest_idx = np.argmax(temperature)

print(f"Coldest star (min temperature): {min_temp} K at row index {coldest_idx} (Mag: {magnitude[coldest_idx]})")
print(f"Hottest star (max temperature): {max_temp} K at row index {hottest_idx} (Mag: {magnitude[hottest_idx]})")

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))

scatter = plt.scatter(
    log_temperature, 
    log_magnitude, 
    c=log_temperature, 
    cmap="coolwarm_r",
    alpha=0.7, 
    label="Stars"
)
plt.colorbar(scatter, label="Temperature (log K)")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest_idx], log_magnitude[brightest_idx], color="black", marker="*", s=300, label="Brightest Star")
plt.scatter(log_temperature[dimmest_idx], log_magnitude[dimmest_idx], color="purple", marker="*", s=300, label="Dimmest Star")
plt.scatter(log_temperature[hottest_idx], log_magnitude[hottest_idx], color="green", marker="*", s=300, label="Hottest Star")
plt.scatter(log_temperature[coldest_idx], log_magnitude[coldest_idx], color="orange", marker="*", s=300, label="Coldest Star")

plt.title("Hertzsprung-Russel Diagram", fontsize=15)
plt.xlabel("Temperature (log K)", fontsize=12)
plt.ylabel("Magnitude", fontsize=12)
plt.legend(loc="best")

plt.tight_layout()
plt.show()
```
