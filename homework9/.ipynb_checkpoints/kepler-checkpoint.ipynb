```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

solar_df = pd.read_csv("../../course_assignments/homework9/solar_system.csv", index_col=0)
```

```python
print("Shape:", solar_df.shape)
print(f"Rows: {solar_df.shape[0]}, Columns: {solar_df.shape[1]}")
print("\nFirst 5 rows:")
print(solar_df.head())
print("\nLast 5 rows:")
print(solar_df.tail())
print("\nColumn names:", solar_df.columns.tolist())
print("\nRow names:", solar_df.index.tolist())
```

```python
orbital_period = solar_df.iloc[0].astype(float)
perihelion = solar_df.iloc[1].astype(float)
aphelion = solar_df.iloc[2].astype(float)
```

```python
#Perihelion: The point in an orbiting object's elliptical path where it is closest to the Sun.
#Aphelion: The point in an orbiting object's elliptical path where it is farthest from the Sun.
#Semi-major axis: Half the length of the longest diameter of an elliptical orbit, representing the average orbital distance from the Sun.
```

```python
semi_major_axis = (perihelion + aphelion) / 2

new_row = pd.Series(semi_major_axis, name="Semi-major axis (AU)").to_frame().T
solar_df = pd.concat([solar_df, new_row])

print("New shape:", solar_df.shape)
print(f"Rows: {solar_df.shape[0]}, Columns: {solar_df.shape[1]}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(semi_major_axis, orbital_period, color="blue", label="Solar System Bodies")
plt.xlabel("Semi-major Axis (AU)")
plt.ylabel("Orbital Period (Earth Years)")
plt.title("Orbital Period vs. Semi-Major Axis")
plt.legend()
plt.show()
```

```python
def keplers_third_law(a, m):
    return m * a ** 1.5

popt, pcov = curve_fit(keplers_third_law, semi_major_axis, orbital_period)
m_fit = popt[0]

print(f"Fit: T = {m_fit:.2f} * a^(1.5)")

residuals = orbital_period - keplers_third_law(semi_major_axis, m_fit)
chi2 = np.sum(residuals**2)
dof = len(semi_major_axis) - len(popt)
chi2_reduced = chi2 / dof

a_smooth = np.linspace(semi_major_axis.min(), semi_major_axis.max(), 200)

plt.figure(figsize=(8, 5))
plt.scatter(semi_major_axis, orbital_period, color="blue", label="Observed Data")
plt.plot(a_smooth, keplers_third_law(a_smooth, m_fit), color="red", label=f"Fitted Model: T = {m_fit:.2f}a^1.5")
plt.title(f"Kepler's Third Law Fit | Reduced Chi-Squared: {chi2_reduced:.2f}")
plt.xlabel("Semi-major Axis (AU)")
plt.ylabel("Orbital Period (Earth Years)")
plt.legend()
plt.show()
```

```python
solar_df.to_csv("kepler.csv")
```
